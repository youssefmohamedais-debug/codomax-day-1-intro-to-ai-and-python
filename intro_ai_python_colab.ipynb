## أولاً: الفرق بين AI و ML و Deep Learning

| المفهوم | التعريف | مثال |
|---|---|---|
| **AI (الذكاء الاصطناعي)** | محاكاة الذكاء البشري لحل المشكلات واتخاذ القرارات | المساعد الصوتي Siri |
| **ML (تعلّم الآلة)** | فرع من AI يتعلّم الأنماط من **البيانات** بدل برمجة قواعد صريحة | تصنيف البريد المزعج Spam |
| **Deep Learning (التعلّم العميق)** | فرع من ML يستخدم **شبكات عصبية** متعددة الطبقات لمعالجة بيانات ضخمة معقّدة | التعرّف على الوجوه والترجمة الفورية |

**العلاقة بينها:** Deep Learning ⊂ ML ⊂ AI (كل تعلّم عميق هو تعلّم آلة، وكل تعلّم آلة هو ذكاء اصطناعي).

## ثانياً: مفاهيم Python الأساسية
### 1) المتغيرات (Variables)

```python
# المتغيرات: حاويات لتخزين البيانات
name = "Sara"            # str  نص
age = 20                 # int  عدد صحيح
height = 1.65            # float عدد عشري
is_student = True        # bool True/False

print("الاسم:", name)
print("العمر:", age)
print("الطول:", height, "متر")
print("طالبة؟", is_student)
print("بعد 5 سنوات سيكون عمرها:", age + 5)
```

### 2) الحلقات (Loops)

```python
# for: تتكرر على عناصر معروفة
print("== عدّاد for ==")
for i in range(1, 6):
    print("الرقم:", i)

# while: تتكرر طالما الشرط صحيح
print("\n== عدّاد while ==")
n = 1
while n <= 5:
    print("الرقم:", n)
    n += 1
```

### 3) الدوال (Functions)

```python
# دالة: كتلة كود قابلة لإعادة الاستخدام
def greet(name):
    return f"أهلاً {name}!"

def add(a, b):
    return a + b

print(greet("أحمد"))
print("3 + 4 =", add(3, 4))
```

### 4) القوائم (Lists)

```python
# قائمة: مجموعة مرتبة وقابلة للتعديل
fruits = ["تفاح", "موز", "برتقال"]
fruits.append("عنب")          # إضافة عنصر
fruits.remove("موز")          # حذف عنصر

print("الفواكه:", fruits)
print("عدد الفواكه:", len(fruits))
print("الأول:", fruits[0], "| الأخير:", fruits[-1])
```

## رابعاً: برامج Python بسيطة (المطلوب تسليمها)

### البرنامج 1: برنامج تعريفي بسيط

```python
# برنامج يطلب اسم المستخدم وعمره ويحسب العمر بعد 10 سنوات
name = input("ما اسمك؟ ")
age = int(input("كم عمرك؟ "))

print("\nأهلاً", name + "!")
print("بعد 10 سنوات سيكون عمرك:", age + 10)
```

### البرنامج 2: زوجي أم فردي (باستخدام دالة وشرط)

```python
# دالة تتحقق إذا كان العدد زوجياً أو فردياً
def check_even_odd(num):
    if num % 2 == 0:
        return "زوجي"
    else:
        return "فردي"

number = int(input("أدخل رقماً: "))
print(f"الرقم {number} هو عدد {check_even_odd(number)}")
```

### البرنامج 3: حلقة تطبع الأرقام من 1 إلى 10 مع مجموعها

```python
# طباعة الأرقام 1..10 وحساب مجموعها باستخدام for
total = 0
for i in range(1, 11):
    print(i, end=" ")
    total += i
print("\nالمجموع =", total)
```

### البرنامج 4: قائمة الدرجات (متوسط وأعلى وأدنى درجة)

```python
# تحليل قائمة درجات الطلاب
grades = [85, 92, 78, 60, 95]

print("الدرجات:", grades)
print("عدد الطلاب:", len(grades))
print("المتوسط:", sum(grades) / len(grades))
print("أعلى درجة:", max(grades))
print("أدنى درجة:", min(grades))
```

### البرنامج 5: لعبة تخمين الرقم (باستخدام while)

```python
# لعبة تخمين الرقم من 1 إلى 10
secret = 7
guess = 0
attempts = 0

while guess != secret:
    guess = int(input("خمّن رقماً بين 1 و 10: "))
    attempts += 1
    if guess < secret:
        print("أكبر!")
    elif guess > secret:
        print("أصغر!")

print(f"أحسنت! خمّنت الرقم {secret} في {attempts} محاولة")
```
